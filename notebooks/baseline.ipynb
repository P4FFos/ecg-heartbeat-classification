```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
)
```

```python
train_data = np.load("../data/processed/train.npz")

print(train_data.files)

for key in train_data.files:
    print(
        key,
        train_data[key].shape,
        train_data[key].dtype
    )
```

```python
train_data = np.load("../data/processed/train.npz")
val_data = np.load("../data/processed/val.npz")

X_train = train_data["X"]
y_train = train_data["y"]
record_train = train_data["record"]

X_val = val_data["X"]
y_val = val_data["y"]
record_val = val_data["record"]

print("Train:")
print("X:", X_train.shape)
print("y:", y_train.shape)

print("\nValidation:")
print("X:", X_val.shape)
print("y:", y_val.shape)

print("\nTraining distribution:")
print(pd.Series(y_train).value_counts())

print("\nValidation distribution:")
print(pd.Series(y_val).value_counts())
```

#Logistic regression

```python
lr = LogisticRegression(
    class_weight="balanced",
    max_iter=5000,
    solver="lbfgs",
    random_state=42
)

lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_val)

print("Logistic Regression")
print("-------------------")
print("Accuracy:", accuracy_score(y_val, y_pred_lr))
print("Macro F1:", f1_score(
    y_val,
    y_pred_lr,
    average="macro"
))

print("\nClassification Report:")
print(classification_report(
    y_val,
    y_pred_lr,
    labels=["N", "S", "V", "F"],
    zero_division=0
))
```

```python
#Random Forest
```

```python
rf = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_val)

print("Random Forest")
print("-------------")
print("Accuracy:", accuracy_score(y_val, y_pred_rf))
print("Macro F1:", f1_score(y_val, y_pred_rf, average="macro"))

print("\nClassification Report:")
print(classification_report(
    y_val,
    y_pred_rf,
    labels=["N", "S", "V", "F"],
    zero_division=0
))
```

```python
#Linear SVM
```

```python
svm = LinearSVC(
    class_weight="balanced",
    C=1.0,
    random_state=42,
    max_iter=5000
)

svm.fit(X_train, y_train)

y_pred_svm = svm.predict(X_val)

print("Linear SVM")
print("----------")
print("Accuracy:", accuracy_score(y_val, y_pred_svm))
print("Macro F1:", f1_score(y_val, y_pred_svm, average="macro"))

print("\nClassification Report:")
print(classification_report(
    y_val,
    y_pred_svm,
    labels=["N", "S", "V", "F"],
    zero_division=0
))
```
